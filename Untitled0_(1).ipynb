```python
import yfinance as yf
import pandas as pd
import numpy as np
from scipy.signal import argrelextrema
import time
from datetime import datetime
import pytz
import os

class LivePaperTrader:
    def __init__(self, ticker="^NSEI", order=5, interval='1m', target_pct=0.01, stoploss_pct=0.005):
        self.ticker = ticker
        self.order = order
        self.interval = interval
        self.position = 0  # 1 for Long, -1 for Short, 0 for Flat
        self.entry_price = 0.0
        self.trade_log = "paper_trade_log.csv"
        self.ist = pytz.timezone('Asia/Kolkata')

        self.target_pct = target_pct
        self.stoploss_pct = stoploss_pct

        print(f"--- Live Paper Trader Started ---")
        print(f"Ticker: {self.ticker} | Target: {target_pct*100}% | SL: {stoploss_pct*100}%")

    def get_latest_signal(self):
        df = yf.download(self.ticker, period='7d', interval=self.interval, progress=False)
        if df.empty: return None, None, {}

        if df.index.tz is None:
            df.index = df.index.tz_localize('UTC').tz_convert(self.ist)
        else:
            df.index = df.index.tz_convert(self.ist)

        df['Support_Pts'] = df.iloc[argrelextrema(df.Low.values, np.less, order=self.order)[0]]['Low']
        df['Resist_Pts'] = df.iloc[argrelextrema(df.High.values, np.greater, order=self.order)[0]]['High']

        s_series = df['Support_Pts'].dropna()
        r_series = df['Resist_Pts'].dropna()

        if len(s_series) < 2 or len(r_series) < 2:
            return 0, df.iloc[-1]['Close'].item(), {}

        last_idx = -1
        current_close = df['Close'].iloc[last_idx].item()
        current_high = df['High'].iloc[last_idx].item()
        current_low = df['Low'].iloc[last_idx].item()
        current_time = df.index[last_idx]

        # Support Projection Logic
        s_p1_idx, s_p1_val = s_series.index[-2], s_series.iloc[-2].item()
        s_p2_idx, s_p2_val = s_series.index[-1], s_series.iloc[-1].item()
        m_s = (s_p2_val - s_p1_val) / (s_p2_idx.value - s_p1_idx.value)
        proj_s = s_p2_val + m_s * (current_time.value - s_p2_idx.value)
        print("support:",s_p1_idx,s_p1_val,s_p2_idx,s_p2_val,proj_s)

        # Resistance Projection Logic
        r_p1_idx, r_p1_val = r_series.index[-2], r_series.iloc[-2].item()
        r_p2_idx, r_p2_val = r_series.index[-1], r_series.iloc[-1].item()
        m_r = (r_p2_val - r_p1_val) / (r_p2_idx.value - r_p1_idx.value)
        proj_r = r_p2_val + m_r * (current_time.value - r_p2_idx.value)
        print("resistance:",r_p1_idx,r_p1_val,r_p2_idx,r_p2_val,proj_r)
        # Pack metadata for logging
        metadata = {
            "S_P1IDX": s_p1_idx, "S_P1": s_p1_val,"S_P2IDX": s_p2_idx, "S_P2": s_p2_val,
            "R_P1IDX": r_p1_idx, "R_P1": r_p1_val,"R_P2IDX": r_p2_idx, "R_P2": r_p2_val,
            "Proj_S": round(proj_s, 2), "Proj_R": round(proj_r, 2)
        }

        signal = 0
        if current_low > proj_r:
            signal = 1  # Bullish Breakout
        elif current_high < proj_s:
            signal = -1 # Bearish Breakout

        return signal, current_close, metadata

    def log_trade(self, action, price, reason="Signal", meta=None):
        timestamp = datetime.now(self.ist).strftime("%Y-%m-%d %H:%M:%S")

        # Base log data
        data = {
            "Timestamp": timestamp,
            "Action": action,
            "Price": round(price, 2),
            "Ticker": self.ticker,
            "Reason": reason
        }

        # Merge metadata (Support/Resistance info) if available
        if meta:
            data.update(meta)

        log_entry = pd.DataFrame([data])

        # Check if file exists to handle header
        file_exists = os.path.isfile(self.trade_log)
        log_entry.to_csv(self.trade_log, mode='a', header=not file_exists, index=False)
        print(f"[{timestamp}] {action} at {price:.2f} | {reason}")

    def run(self):
        while True:
            try:
                signal, price, meta = self.get_latest_signal()
                if signal is None:
                    continue

                # --- EXIT LOGIC ---
                if self.position == 1:
                    if price >= self.entry_price * (1 + self.target_pct):
                        self.log_trade("EXIT/LONG", price, "TARGET HIT", meta)
                        self.position = 0
                    elif price <= self.entry_price * (1 - self.stoploss_pct):
                        self.log_trade("EXIT/LONG", price, "STOP LOSS HIT", meta)
                        self.position = 0

                elif self.position == -1:
                    if price <= self.entry_price * (1 - self.target_pct):
                        self.log_trade("EXIT/SHORT", price, "TARGET HIT", meta)
                        self.position = 0
                    elif price >= self.entry_price * (1 + self.stoploss_pct):
                        self.log_trade("EXIT/SHORT", price, "STOP LOSS HIT", meta)
                        self.position = 0

                # --- ENTRY LOGIC ---
                if self.position == 0:
                    if signal == 1:
                        self.log_trade("BUY/LONG", price, "Breakout", meta)
                        self.position = 1
                        self.entry_price = price
                    elif signal == -1:
                        self.log_trade("SELL/SHORT", price, "Breakout", meta)
                        self.position = -1
                        self.entry_price = price

                status = "FLAT" if self.position == 0 else ("LONG" if self.position == 1 else "SHORT")
                print(f"[{datetime.now(self.ist).strftime('%H:%M:%S')}] Price: {price:.2f} | Pos: {status}", end='\r')

                time.sleep(30)

            except Exception as e:
                print(f"\nError: {e}")
                time.sleep(10)

if __name__ == "__main__":
    # Note: Target/SL adjusted slightly for realism, your original values were extremely tight
    bot = LivePaperTrader(ticker="^NSEI", interval='1m', target_pct=0.0005, stoploss_pct=0.0003)
    bot.run()
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
from scipy.signal import argrelextrema

class SRBacktester:
    def __init__(self, ticker="^NSEI", order=5):
        self.ticker = ticker
        self.order = order

    def get_signals(self, interval='5m', period='1mo'):
        df = yf.download(self.ticker, period=period, interval=interval)
        if df.empty: return pd.DataFrame()

        # Find all peaks and troughs in one go
        df['Support_Pts'] = df.iloc[argrelextrema(df.Low.values, np.less, order=self.order)[0]]['Low']
        df['Resist_Pts'] = df.iloc[argrelextrema(df.High.values, np.greater, order=self.order)[0]]['High']

        df['Signal'] = 0  # 1 for Buy, -1 for Sell

        # We start from where we have at least 2 points for S and R
        # This is a simplified loop for backtesting logic
        for i in range(len(df)):
            if i < self.order*4: continue

            # Look at data available up to 'i' (to avoid look-ahead bias)
            lookback = df.iloc[:i]
            s_series = lookback['Support_Pts'].dropna()
            r_series = lookback['Resist_Pts'].dropna()

            if len(s_series) >= 2 and len(r_series) >= 2:
                # Support Ray projection
                s_p1_idx, s_p1_val = s_series.index[-2], s_series.iloc[-2].item()
                s_p2_idx, s_p2_val = s_series.index[-1], s_series.iloc[-1].item()

                # Math: slope = (y2-y1)/(x2-x1)
                m_s = (s_p2_val - s_p1_val) / (s_p2_idx.value - s_p1_idx.value)
                proj_s = s_p2_val + m_s * (df.index[i].value - s_p2_idx.value)

                # Resistance Ray projection
                r_p1_idx, r_p1_val = r_series.index[-2], r_series.iloc[-2].item()
                r_p2_idx, r_p2_val = r_series.index[-1], r_series.iloc[-1].item()

                m_r = (r_p2_val - r_p1_val) / (r_p2_idx.value - r_p1_idx.value)
                proj_r = r_p2_val + m_r * (df.index[i].value - r_p2_idx.value)

                # Check crossover
                res_current_close = df['Low'].iloc[i].item()
                sup_current_close = df['High'].iloc[i].item()

                if res_current_close > proj_r:
                    df.iloc[i, df.columns.get_loc('Signal')] = 1
                elif sup_current_close < proj_s:
                    df.iloc[i, df.columns.get_loc('Signal')] = -1

        return df

# Run and Analyze
tester = SRBacktester()
results = tester.get_signals(interval='15m', period='1mo') # Changed period to '1mo' to satisfy 60-day limit for 15m interval
print(results)
# Simple Performance Metric
#results['Returns'] = results['Close'].pct_change()
#results['Strategy_Returns'] = results['Signal'].shift(1) * results['Returns']
#print(f"Cumulative Strategy Return: {(1 + results['Strategy_Returns']).prod() - 1:.2%}")
# --- Realistic Performance Calculation ---

# 1. Define your costs (0.05% for slippage + taxes)
cost_per_trade = 0.0005

# 2. Identify where trades actually happened
# A trade occurs when the signal changes (e.g., from 0 to 1, or 1 to -1)
results['Trade_Signal'] = results['Signal'].diff().fillna(0)
results['Has_Traded'] = results['Trade_Signal'] != 0

# 3. Calculate Raw Strategy Returns
results['Returns'] = results['Close'].pct_change()
results['Strategy_Returns_Raw'] = results['Signal'].shift(1) * results['Returns']

# 4. Subtract Costs only on bars where a trade was executed
results['Strategy_Returns_Net'] = results['Strategy_Returns_Raw']
results.loc[results['Has_Traded'], 'Strategy_Returns_Net'] -= cost_per_trade

# 5. Compare Results
raw_final = (1 + results['Strategy_Returns_Raw']).prod() - 1
net_final = (1 + results['Strategy_Returns_Net']).prod() - 1
print(results['Has_Traded'].sum())
print(f"Raw Return: {raw_final:.2%}")
print(f"Net Return (After Costs): {net_final:.2%}")
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
from scipy.signal import argrelextrema
import time
from datetime import datetime
import pytz

class LivePaperTrader:
    def __init__(self, ticker="^NSEI", order=10, interval='1m', target_pct=0.01, stoploss_pct=0.005):
        self.ticker = ticker
        self.order = order
        self.interval = interval
        self.position = 0  # 1 for Long, -1 for Short, 0 for Flat
        self.entry_price = 0.0
        self.trade_log = "paper_trade_log.csv"
        self.ist = pytz.timezone('Asia/Kolkata')

        # Risk Management Parameters
        self.target_pct = target_pct
        self.stoploss_pct = stoploss_pct

        print(f"--- Live Paper Trader Started ---")
        print(f"Ticker: {self.ticker} | Target: {target_pct*100}% | SL: {stoploss_pct*100}%")

    def get_latest_signal(self):
        df = yf.download(self.ticker, period='7d', interval=self.interval, progress=False)
        if df.empty: return None, None

        if df.index.tz is None:
            df.index = df.index.tz_localize('UTC').tz_convert(self.ist)
        else:
            df.index = df.index.tz_convert(self.ist)

        df['Support_Pts'] = df.iloc[argrelextrema(df.Low.values, np.less, order=self.order)[0]]['Low']
        df['Resist_Pts'] = df.iloc[argrelextrema(df.High.values, np.greater, order=self.order)[0]]['High']

        s_series = df['Support_Pts'].dropna()
        r_series = df['Resist_Pts'].dropna()

        if len(s_series) < 2 or len(r_series) < 2:
            return 0, df.iloc[-1]['Close'].item()

        last_idx = -1
        current_close = df['Close'].iloc[last_idx].item()
        current_high = df['High'].iloc[last_idx].item()
        current_low = df['Low'].iloc[last_idx].item()
        current_time = df.index[last_idx]

        # Support/Resistance Projection Logic
        s_p1_idx, s_p1_val = s_series.index[-2], s_series.iloc[-2].item()
        s_p2_idx, s_p2_val = s_series.index[-1], s_series.iloc[-1].item()
        m_s = (s_p2_val - s_p1_val) / (s_p2_idx.value - s_p1_idx.value)
        proj_s = s_p2_val + m_s * (current_time.value - s_p2_idx.value)
        print("support:",s_p1_idx,s_p1_val,s_p2_idx,s_p2_val,proj_s)


        r_p1_idx, r_p1_val = r_series.index[-2], r_series.iloc[-2].item()
        r_p2_idx, r_p2_val = r_series.index[-1], r_series.iloc[-1].item()
        m_r = (r_p2_val - r_p1_val) / (r_p2_idx.value - r_p1_idx.value)
        proj_r = r_p2_val + m_r * (current_time.value - r_p2_idx.value)
        print("resistance:",r_p1_idx,r_p1_val,r_p2_idx,r_p2_val,proj_r)
        signal = 0
        if current_low > proj_r:
            signal = 1  # Bullish Breakout
        elif current_high < proj_s:
            signal = -1 # Bearish Breakout

        return signal, current_close

    def log_trade(self, action, price, reason="Signal"):
        timestamp = datetime.now(self.ist).strftime("%Y-%m-%d %H:%M:%S")
        log_entry = pd.DataFrame([{
            "Timestamp": timestamp,
            "Action": action,
            "Price": round(price, 2),
            "Ticker": self.ticker,
            "Reason": reason
        }])
        log_entry.to_csv(self.trade_log, mode='a', header=not pd.io.common.file_exists(self.trade_log), index=False)
        print(f"[{timestamp}] {action} at {price:.2f} ({reason})")

    def run(self):
        while True:
            try:
                signal, price = self.get_latest_signal()
                if signal is None:
                    continue

                # --- EXIT LOGIC (SL / Target) ---
                if self.position == 1: # Long
                    if price >= self.entry_price * (1 + self.target_pct):
                        self.log_trade("EXIT/LONG", price, "TARGET HIT")
                        self.position = 0
                    elif price <= self.entry_price * (1 - self.stoploss_pct):
                        self.log_trade("EXIT/LONG", price, "STOP LOSS HIT")
                        self.position = 0

                elif self.position == -1: # Short
                    if price <= self.entry_price * (1 - self.target_pct):
                        self.log_trade("EXIT/SHORT", price, "TARGET HIT")
                        self.position = 0
                    elif price >= self.entry_price * (1 + self.stoploss_pct):
                        self.log_trade("EXIT/SHORT", price, "STOP LOSS HIT")
                        self.position = 0
                else:
                    print(f"Holding Position: {self.position} | Current Price: {price:.2f}")

                # --- ENTRY LOGIC (Signals) ---
                if self.position == 0:
                    if signal == 1:
                        self.log_trade("BUY/LONG", price, "Breakout")
                        self.position = 1
                        self.entry_price = price
                    elif signal == -1:
                        self.log_trade("SELL/SHORT", price, "Breakout")
                        self.position = -1
                        self.entry_price = price

                # Console Update
                status = "FLAT" if self.position == 0 else ("LONG" if self.position == 1 else "SHORT")
                print(f"[{datetime.now(self.ist).strftime('%H:%M:%S')}] Price: {price:.2f} | Pos: {status}", end='\r')

                time.sleep(30)

            except Exception as e:
                print(f"\nError: {e}")
                time.sleep(10)

if __name__ == "__main__":
    # Example: 0.5% Target, 0.25% Stop Loss
    bot = LivePaperTrader(ticker="^NSEI", interval='1m', target_pct=0.000025, stoploss_pct=0.000015)
    bot.run()
```
